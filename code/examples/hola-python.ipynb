```python
import pandas as pd
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt

datos = pd.DataFrame({"x": [1, 2, 3, 4, 5], "y": [2, 4, 5, 4, 6]})
modelo = smf.ols("y ~ x", data=datos).fit()
modelo.params
```

```python
datos.plot.scatter(x="x", y="y")
plt.show()
```
